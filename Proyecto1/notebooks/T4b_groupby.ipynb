## Objetivo

Calcular estadísticas de supervivencia por categoría con `groupby`, siguiendo el contrato del equipo (sin bucles `for` sobre las filas).

```python
import pandas as pd
import numpy as np

# Carga del dataset desde la carpeta Data del repositorio.
# Si no se encuentra (por ejemplo al abrir el notebook en Colab), se descarga
# de la misma fuente original, así que los datos son idénticos.
try:
    df = pd.read_csv("../Data/titanic.csv")
except FileNotFoundError:
    df = pd.read_csv("https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv")

df.shape
```

Aplicamos la base común del contrato (tratamiento de nulos, ver T2a).

```python
# --- BASE COMÚN (tratamiento de nulos acordado en el contrato, ver T2a) ---
df["age"] = df["age"].fillna(df["age"].median())
df["embarked"] = df["embarked"].fillna(df["embarked"].mode()[0])
df["embark_town"] = df["embark_town"].fillna(df["embark_town"].mode()[0])
df = df.drop(columns="deck")
```

## 1. Tasa de supervivencia por sexo

```python
# Media de survived (0/1) por sexo = proporción de supervivientes, en %
tasa_sexo = (df.groupby("sex")["survived"].mean() * 100).round(2)
tasa_sexo
```

## 2. Tasa de supervivencia por clase

```python
tasa_clase = (df.groupby("pclass")["survived"].mean() * 100).round(2)
tasa_clase
```

## 3. Tasa de supervivencia por sexo y clase combinados
Agrupamos por las dos variables a la vez para ver si combinarlas cambia el panorama respecto a mirarlas por separado. Con `unstack()` lo mostramos como tabla (filas: sexo, columnas: clase).

```python
tasa_sexo_clase = (df.groupby(["sex", "pclass"])["survived"].mean() * 100).round(2)
tasa_sexo_clase.unstack()
```

## 4. Tarifa y edad media por clase

```python
# Varias estadísticas a la vez: media de tarifa y de edad por clase
df.groupby("pclass")[["fare", "age"]].mean().round(2)
```

## Hallazgo

Por separado, el sexo marca más diferencia que la clase: sobrevivió el 74,20% de las mujeres frente al 18,89% de los hombres, mientras que por clase el rango va del 62,96% (1ª) al 24,24% (3ª). Al cruzar ambas variables aparece algo que no se ve por separado: entre las mujeres, la clase casi no importa en 1ª y 2ª (96,81% y 92,11%) y la caída llega en 3ª (50,00%); entre los hombres, la supervivencia es baja en todas las clases (36,89%, 15,74% y 13,54%). En puntos porcentuales, pasar de 1ª a 3ª clase supuso una caída de 46,8 puntos para las mujeres y de 23,4 para los hombres. Además, la tarifa media en 1ª clase (84,15) es más de 6 veces la de 3ª (13,68), lo que confirma que clase y poder adquisitivo van de la mano.
